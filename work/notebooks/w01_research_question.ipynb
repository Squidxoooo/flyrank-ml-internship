## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
## 2. The question

**Research question:**  
Which visible pages under-capture clicks relative to their search-position peers and therefore deserve review first?

**Unit of analysis:**  
One content page over an observed 90-day performance window.

**Decision:**  
Which pages should a reviewer investigate first for a possible CTR or engagement improvement?

**Action:**  
For the highest-priority pages, recommend actions such as reviewing title/meta presentation, checking intent alignment, improving snippet structure, reviewing on-page engagement, or monitoring the page when the evidence is weak.

**Expected output:**  
A ranked opportunity list containing the page identifier, impression volume, observed CTR, expected CTR for its position group, CTR gap, reason codes, and a suggested action.

**Cost of getting it wrong:**  
The main risk is wasting limited review time on low-volume or noisy pages, or incorrectly treating a low CTR as a content problem when the difference is mainly explained by search position or demand.
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# ML-02 — Starter data evidence

import pandas as pd

# Load starter dataset
df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

# Keep pages with at least 100 impressions
visible = df[df["impressions_90d"] >= 100].copy()

# Mean CTR by position tier
tier_order = ["top_3", "page_1", "striking", "page_3_5", "deep"]

ctr_by_position = (
    visible.groupby("position_tier")["ctr"]
    .mean()
    .reindex(tier_order)
)

print("Mean CTR by position tier (impressions >= 100):")
print(ctr_by_position.round(4))

print()
print(f"Pages after volume filter: {len(visible):,}")
print(f"Page 1 mean CTR: {ctr_by_position['page_1']:.4f}%")
print(f"Deep-position mean CTR: {ctr_by_position['deep']:.4f}%")
```
